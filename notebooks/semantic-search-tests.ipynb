# Semantic search tests

```python
import os
from dotenv import load_dotenv
from vector_search.semantic_search import Neo4jConnection, semantic_search_with_context_parallel
```

## Database connection

```python
load_dotenv()

conn = Neo4jConnection(
    uri=os.getenv("NEO4J_URI"),
    user=os.getenv("NEO4J_USER"),
    password=os.getenv("NEO4J_PASSWORD")
)
```

```python
question = "¿Que se sabe de la Granja 2?"
top_k_nodos = 3
profundidad = 2
numero_workers = 10  # Ajusta este valor según tus recursos y límites de la API

resultados = semantic_search_with_context_parallel(
    neo4j=conn,
    question=question,
    top_k=top_k_nodos,
    depth=profundidad,
    max_workers=numero_workers
)

# Procesa los resultados
print(resultados)
```

```python
print("\nRelaciones expandidas:")
for triple in resultado["expanded_context"]:
    origen = triple["n"].get("id", "??")
    destino = triple["m"].get("id", "??")
    relacion = triple["r"]

    tipo = relacion.type if hasattr(relacion, "type") else "REL"
    props = dict(relacion.items()) if hasattr(relacion, "items") else {}

    print(f"{origen} -[{tipo}]-> {destino}")
    if props:
        print("  Atributos de la relación:")
        for k, v in props.items():
            print(f"    {k}: {v}")
```
